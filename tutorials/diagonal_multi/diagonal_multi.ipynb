# Diagonal integration: [multiple RNA, multiple ATAC]

Diagonal integration combines RNA and ATAC measured in different cells, with no pairing between them. This tutorial is for data in several batches: three RNA batches and three ATAC batches. It runs iNMF and online_iNMF on `D37mini`, 3,000 RNA and 3,000 ATAC profiles of the benchmark dataset `D37`, then on data in your own format.

iNMF, online_iNMF and scJoint read ATAC as gene-activity scores, made beforehand with a tool such as Signac or ArchR. GLUE reads the peak matrix. The dataset holds both.

Methods run on Linux. On macOS or Windows, [open this notebook in Colab](https://colab.research.google.com/github/DSichang/scMultiBench/blob/main/notebooks/tutorial_diagonal_multi.ipynb).

## 2. Download the data and the environments

`mtb.data.fetch` downloads `D37mini` (12 MB) once. Each method runs in its own environment. `mtb.env.install` downloads the environments for iNMF and online_iNMF, with no conda needed. The download is 0.9 GB.

<details>
<summary>Details</summary>

`state` is `PACKED` for an environment downloaded now and `have` for one that was already there. Without `dry_run=False`, `mtb.env.install` downloads nothing and returns the plan with its sizes.

Environments go to `mtb.config.DEFAULT.envs_dir`. To use another disk, set it before this cell.

From a terminal, `multibench env install --methods iNMF,online_iNMF --packed --run` does the same.

</details>

```python
import pandas as pd
import multibench as mtb

mtb.data.fetch("D37mini")
```

```python
METHODS = ["iNMF", "online_iNMF"]
envs = mtb.env.install(METHODS, dry_run=False)
pd.DataFrame(envs)[["env", "methods", "state"]]
```

## 3. Run the methods

`run_all` runs each method on `D37mini` and scores its output with the scIB metrics and the classification scores (OCA, ACA, Spec, Sens, F1). `res.summary` has one row per method with its status, run time and scores. Higher is better for every metric.

<details>
<summary>Details</summary>

Each method writes an embedding: a table of numbers with one row per cell. `run_all` scores it against the cell-type labels of `D37mini`.

The clustering metrics `ARI`, `NMI`, `ASW`, `iASW`, `iF1` and `cLISI` measure how well the embedding separates the cell types. The batch metrics `ASW_batch`, `GC` and `iLISI` measure how well the batches mix. They appear only when the data has several batches.

`res.failures` says why a method failed. `params={"Method": {"key": value}}` sets a method's parameters, and `mtb.params_for` lists them. Many methods take none.

`mtb.load_batch("out/D37mini")` reloads these results later without running anything.

</details>

```python
res = mtb.run_all("D37mini", "diagonal", methods=METHODS,
                  out_dir="out/D37mini")
res.summary
```

## 4. Plot

`res.plot()` draws the scores as a bubble table. Circle size shows the rank within a column, and bigger is better. The fill compares a value with the other rows in the same column.

<details>
<summary>Details</summary>

The lightest fill is the lowest value in this figure, not zero.

Metrics are grouped by family: blue for dimension reduction and clustering, green for batch correction, pink for classification. Each family starts with an `Overall` bar. Its length and colour both show the family score.

A column whose rows all hold the same value is drawn grey, and a warning names it. A figure of one method is grey everywhere.

</details>

```python
res.plot()
```

## 5. Your own data

Your data needs raw RNA counts and ATAC as one `.h5ad` file per batch, with a cell type for each cell. A batch is a donor, a sample or an experiment. `mtb.io.export_dataset` with `batch_index=` writes one RNA batch and one ATAC batch per call. iNMF and online_iNMF read ATAC as gene-activity scores, and online_iNMF needs about 5,000 cells in total. Gene-activity scores are an ATAC matrix with one column per gene, as Signac or ArchR compute it.

Starting from something else? See the recipes for [labels in a CSV file](https://dsichang.github.io/scMultiBench/tutorials/run/#recipe-labels-csv) and [no labels](https://dsichang.github.io/scMultiBench/tutorials/run/#recipe-no-labels).

<details>
<summary>Details: export</summary>

The names in quotes say where your file holds each part. `"X"` is the main matrix, `"obsm:protein"` the matrix `obsm["protein"]`, `"layer:counts"` the layer `layers["counts"]`, and `"obs:celltype"` the column `obs["celltype"]`, which `"celltype"` alone names too. A wrong name after `obsm:`, `layer:` or `obs:` raises an error that lists the names the file has.

The RNA batches and the ATAC batches are numbered separately. RNA batch 1 and ATAC batch 1 need not come from the same sample. The loop takes as many batches as the list `rna` holds.

For ATAC as a peak matrix, pass `atac_kind="peak"`. GLUE reads the peak matrix.

`NAME` is the dataset name for `scan` and `run_all`, and `mydata` is the folder that holds it (`data_path`). With a new value of `NAME`, the folder holds the files of your data only.

The check lists the methods you name. Without `methods=`, it lists every method that reads the modalities in the folder. Without `per_method=True`, it has one row per method variant and every column, such as the command of each method.

`export_dataset` raises `FileExistsError` when the folder holds files of other cells that the call does not write. `overwrite=True` does not remove them. Give `NAME` a new value. A call with `batch_index=` checks the files of its own batch number. Files of other batch numbers stay, unless `n_batches=` says that the data has fewer batches.

`overwrite=True` replaces the files of an earlier run of this cell. Without it, `export_dataset` raises `FileExistsError` rather than replace a file.

</details>

The cells below read your data, write it as a dataset folder, check the folder and run the same methods on it. So that the notebook runs as it is, the next cell writes six small example files, `my_rna1.h5ad` to `my_rna3.h5ad` and `my_atac1.h5ad` to `my_atac3.h5ad`, from the first 900 RNA profiles and the first 900 ATAC profiles of each batch of `D37mini`. It leaves a file of your own with one of those names as it is.

```python
mtb.data.write_example("D37mini")
```

Read your files. For your own data, change every file name marked `# <-`. A name that you leave is read from the example. On Colab, upload the files first: click the folder icon on the left, then the upload icon.

```python
import anndata as ad

rna1 = ad.read_h5ad("my_rna1.h5ad")     # <- RNA batch 1: your file
rna2 = ad.read_h5ad("my_rna2.h5ad")     # <- RNA batch 2: your file
rna3 = ad.read_h5ad("my_rna3.h5ad")     # <- RNA batch 3: your file
atac1 = ad.read_h5ad("my_atac1.h5ad")   # <- ATAC batch 1: your file
atac2 = ad.read_h5ad("my_atac2.h5ad")   # <- ATAC batch 2: your file
atac3 = ad.read_h5ad("my_atac3.h5ad")   # <- ATAC batch 3: your file
print(rna1)
print(atac1)
```

Write the dataset folder. Change each name in quotes marked `# <-`: give `NAME` a new value, and write the names that your files use for the other parts. The printout above lists the names in two of your files. The other files need the same names. Write a name listed after `obs:` with that prefix, as `"obs:celltype"`. The loop writes one RNA batch and one ATAC batch per call, and `n_batches` says how many batches the data has.

```python
NAME = "MYDIAG_BATCHES"        # <- a new name for your own data
rna = [rna1, rna2, rna3]       # <- your RNA batches
atac = [atac1, atac2, atac3]   # <- your ATAC batches
for b in range(len(rna)):      # one call per batch
    mtb.io.export_dataset(rna[b], "mydata/" + NAME,
                          rna="X",                    # <- the RNA counts
                          atac=atac[b],               # the ATAC batch
                          atac_kind="gene_activity",  # <- or "peak"
                          labels="obs:celltype",      # <- the cell-type column
                          category="diagonal", batch_index=b + 1,
                          n_batches=len(rna), overwrite=True)
```

Check the folder before you run anything. `runnable` says whether a method can run on your data on this computer, and `reason` says why not. The first two lines above the table say what the folder holds and how many cells each file has. Compare them with your data. A further line is about one method: why it cannot run, or what it does not read or needs from your data.

```python
mtb.scan(NAME, "diagonal", methods=METHODS, data_path="mydata",
         per_method=True)
```

Run the same methods on the folder:

```python
mine = mtb.run_all(NAME, "diagonal", methods=METHODS, data_path="mydata",
                   out_dir="out/" + NAME)
mine.summary
```

```python
mine.plot()
```

## Troubleshooting

`res.failures` lists each method that failed or was skipped. Its `error` column ends with the method's error output.

<details>
<summary>Details</summary>

| symptom | fix |
|---|---|
| `env.install` refuses on macOS or Windows | methods run only on Linux: use Colab or a Linux machine |
| a method needs an NVIDIA GPU | choose a GPU runtime on Colab, or a machine with a GPU |
| a method fails on a small dataset | `mtb.scan` without `per_method=True` names what the method needs from the data's size in its `caveat` column |
| a warning that values are not whole numbers | export raw counts, for example with `rna="layer:counts"` |
| `... matrix/data as cells x features` | the matrix is transposed: export it again with `mtb.io.export_dataset` |
| a method times out | raise `timeout=` in `run_all` |
| low `label_order_confidence` | several label files fit the cell count: check `label_order_candidates` in `res.results` |
| batch metrics use the wrong batches | `res.rescore(batch=my_vector)` scores again without running the methods |

</details>
