# Mosaic integration: [RNA, RNA + ATAC, ATAC], every method

This notebook runs every method that has a variant for mosaic [RNA, RNA + ATAC, ATAC] on `D45mini`: Cobolt, MultiVI, Multigrate, SMILE, StabMap and scMoMaT. The [tutorial](https://dsichang.github.io/scMultiBench/tutorials/mosaic_rna_atac/) explains each step and runs StabMap and scMoMaT only.

Methods run on Linux. The 6 environments download 7.8 GB on a computer without a GPU and 8.9 GB on one with a GPU.

## 1. Install

pip may then print a note about restarting the kernel: no restart is needed, because the pins keep the numpy and pandas already installed. On Colab, choose a T4 GPU runtime first (Runtime -> Change runtime type), because changing it later loses the downloads.

```python
import numpy, pandas
pins = f"numpy=={numpy.__version__} pandas=={pandas.__version__}"
%pip install -q "multibench-sc>=0.7.0" {pins}
```

## 2. Download the data and the environments

```python
import multibench as mtb
import pandas as pd

pd.set_option("display.max_columns", None)   # show every column of the tables

mtb.data.fetch("D45mini")
METHODS = ["Cobolt", "MultiVI", "Multigrate", "SMILE", "StabMap", "scMoMaT"]
envs = mtb.env.install(METHODS)
```

## 3. Run the methods

```python
res = mtb.run("D45mini", "mosaic", methods=METHODS, out_dir="out/D45mini_all")
res.summary
```

## 4. Plot

`res.plot()` draws the scores as a bubble table. Circle size ranks the rows within each column and colour shows the value scaled per column, so a small gap can look large. `res.summary` has the values.

```python
res.plot()
```
