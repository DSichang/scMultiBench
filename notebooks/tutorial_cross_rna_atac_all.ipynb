# Cross integration: Multiple RNA + ATAC, every method

This notebook runs every method that has a variant for cross Multiple RNA + ATAC on `D56mini`: MOFA2, Multigrate, StabMap, UINMF, UnitedNet, scMDC, scMM and scMoMaT. The [tutorial](https://dsichang.github.io/scMultiBench/tutorials/cross_rna_atac/) explains each step and runs StabMap and scMM only.

Methods run on Linux. The environments are 6 envs, 23.3 GB on a computer without a GPU and 23.6 GB on one with a GPU.

## 1. Install

This cell installs `multibench-sc`. On Colab, choose a T4 GPU runtime first (Runtime -> Change runtime type), because changing it later loses the downloads.

```python
import numpy, pandas
pins = f"numpy=={numpy.__version__} pandas=={pandas.__version__}"
%pip install -q "multibench-sc>=0.5.0" {pins}
```

## 2. Download the data and the environments

```python
import multibench as mtb

mtb.data.fetch("D56mini")
METHODS = ["MOFA2", "Multigrate", "StabMap", "UINMF", "UnitedNet", "scMDC", "scMM",
           "scMoMaT"]
envs = mtb.env.install(METHODS)
```

## 3. Run the methods

```python
res = mtb.run("D56mini", "cross", methods=METHODS,
                  out_dir="out/D56mini_all")
res.summary
```

## 4. Plot

`res.plot()` draws the scores as a bubble table. The fills and Overall bars rank the rows within each column, so a small gap can look large: `res.summary` has the values.

```python
res.plot()
```
