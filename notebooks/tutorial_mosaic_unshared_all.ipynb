# Mosaic integration: Mixed, without shared modality, every method

This notebook runs every method that has a variant for mosaic Mixed, without shared modality on `D49mini`: Multigrate, StabMap and scMoMaT. The [tutorial](https://dsichang.github.io/scMultiBench/tutorials/mosaic_unshared/) explains each step and runs StabMap and scMoMaT only.

Methods run on Linux. The environments are 3 envs, 2.5 GB on a computer without a GPU and 5.9 GB on one with a GPU.

## 1. Install

This cell installs `multibench-sc`. On Colab, choose a T4 GPU runtime first (Runtime -> Change runtime type), because changing it later loses the downloads.

```python
import numpy, pandas
pins = f"numpy=={numpy.__version__} pandas=={pandas.__version__}"
%pip install -q "multibench-sc>=0.5.0" {pins}
```

## 2. Download the data and the environments

```python
import multibench as mtb

mtb.data.fetch("D49mini")
METHODS = ["Multigrate", "StabMap", "scMoMaT"]
envs = mtb.env.install(METHODS)
```

## 3. Run the methods

```python
res = mtb.run("D49mini", "mosaic", methods=METHODS,
                  out_dir="out/D49mini_all")
res.summary
```

## 4. Plot

`res.plot()` draws the scores as a bubble table. The bubbles and Overall bars rank the rows within each column, so a small gap can look large: `res.summary` has the values.

```python
res.plot()
```
