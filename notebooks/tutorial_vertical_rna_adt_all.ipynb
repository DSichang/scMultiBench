# Vertical integration: RNA + ADT, every method

This notebook runs every method that has a variant for vertical RNA + ADT on `D11`: Concerto, MOFA2, Matilda, Multigrate, Seurat_WNN, UINMF, VIMCCA, moETM, scMDC, scMM, scMSI, scMoMaT, sciPENN and totalVI. The [tutorial](https://dsichang.github.io/scMultiBench/tutorials/vertical_rna_adt/) explains each step and runs Matilda and sciPENN only.

Methods run on Linux. The environments are 13 envs, 24.5 GB on a computer without a GPU and 31.9 GB on one with a GPU.

## 1. Install

The pins keep the numpy and pandas already installed, so Colab needs no restart. On Colab, choose a T4 GPU runtime first (Runtime -> Change runtime type), because changing it later loses the downloads.

```python
import numpy, pandas
pins = f"numpy=={numpy.__version__} pandas=={pandas.__version__}"
%pip install -q "multibench-sc>=0.5.2" {pins}
```

## 2. Download the data and the environments

```python
import multibench as mtb

mtb.data.fetch("D11")
METHODS = ["Concerto", "MOFA2", "Matilda", "Multigrate", "Seurat_WNN", "UINMF",
           "VIMCCA", "moETM", "scMDC", "scMM", "scMSI", "scMoMaT", "sciPENN",
           "totalVI"]
envs = mtb.env.install(METHODS)
```

## 3. Run the methods

Seurat_WNN returns a neighbour graph, not an embedding. The package scores embeddings, so its row has the status `ran` and no scores, and the figure leaves it out.

```python
res = mtb.run("D11", "vertical", methods=METHODS, out_dir="out/D11_all")
res.summary
```

## 4. Plot

`res.plot()` draws the scores as a bubble table. Circle size ranks the rows within each column and colour shows the value scaled per column, so a small gap can look large. `res.summary` has the values.

```python
res.plot()
```
