# Vertical integration: RNA + ADT + ATAC, every method

This notebook runs every method that has a variant for vertical RNA + ADT + ATAC on `D22mini`: MOFA2, Matilda, Multigrate, UINMF and scMoMaT. The [tutorial](https://dsichang.github.io/scMultiBench/tutorials/vertical_rna_adt_atac/) explains each step and runs Matilda and UINMF only.

Methods run on Linux. The environments are 5 envs, 12.7 GB on a computer without a GPU and 18.5 GB on one with a GPU.

## 1. Install

This cell installs `multibench-sc`. On Colab, choose a T4 GPU runtime first (Runtime -> Change runtime type), because changing it later loses the downloads.

```python
import numpy, pandas
pins = f"numpy=={numpy.__version__} pandas=={pandas.__version__}"
%pip install -q "multibench-sc>=0.5.1" {pins}
```

## 2. Download the data and the environments

```python
import multibench as mtb

mtb.data.fetch("D22mini")
METHODS = ["MOFA2", "Matilda", "Multigrate", "UINMF", "scMoMaT"]
envs = mtb.env.install(METHODS)
```

## 3. Run the methods

```python
res = mtb.run("D22mini", "vertical", methods=METHODS,
              out_dir="out/D22mini_all")
res.summary
```

## 4. Plot

`res.plot()` draws the scores as a bubble table. Circle size ranks the rows within each column and the colour scales each column, so a small gap can look large: `res.summary` has the values.

```python
res.plot()
```
