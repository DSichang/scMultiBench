# Mosaic integration: RNA + ADT, every method

This notebook runs every method that has a variant for mosaic RNA + ADT on `D38mini`: Multigrate. The [tutorial](https://dsichang.github.io/scMultiBench/tutorials/mosaic_rna_adt/) explains each step and runs Multigrate only.

Methods run on Linux. The environments are 1 env, 4.4 GB to download on a CPU host, 4.4 GB on a GPU host.

## 2. Download the data and the environments

```python
import pandas as pd
import multibench as mtb

mtb.data.fetch("D38mini")
```

```python
METHODS = ["Multigrate"]
envs = mtb.env.install(METHODS, dry_run=False)
pd.DataFrame(envs)[["env", "methods", "state"]]
```

## 3. Run the method

```python
res = mtb.run_all("D38mini", "mosaic", methods=METHODS,
                  out_dir="out/D38mini_all")
res.summary
```

## 4. Plot

`res.plot()` draws the scores as a bubble table. Circle size shows the rank within a column, and bigger is better. The fill compares a value with the other rows in the same column.

```python
res.plot()
```
