# Vertical integration: RNA + ADT + ATAC

Vertical integration combines modalities measured in the same cells. This tutorial is for RNA, surface protein and ATAC from the same cells. It runs Matilda and UINMF on `D22mini`, 3,000 cells of the benchmark dataset `D22`, and then on data in your own format.

Methods run on Linux. On macOS or Windows, [open this notebook in Colab](https://colab.research.google.com/github/DSichang/scMultiBench/blob/main/notebooks/tutorial_vertical_rna_adt_atac.ipynb).

## 1. Install

This cell installs `multibench-sc`. On Colab, choose a T4 GPU runtime first (Runtime -> Change runtime type), because changing it later loses the downloads.

```python
import numpy, pandas
pins = f"numpy=={numpy.__version__} pandas=={pandas.__version__}"
%pip install -q "multibench-sc>=0.5.0" {pins}
```

## 2. Download the data and the environments

`mtb.data.fetch` downloads `D22mini` (8 MB) once. `mtb.env.install` downloads the environments for Matilda and UINMF. The download is 1.5 GB on a computer without a GPU and 4.0 GB on one with a GPU.

```python
import multibench as mtb

mtb.data.fetch("D22mini")
METHODS = ["Matilda", "UINMF"]
envs = mtb.env.install(METHODS)
```

## 3. Run the methods

`run` runs each method on `D22mini` and scores its output with the scIB metrics. The clustering metrics score how well the embedding's clusters and neighbourhoods match the cell types, and how separate the clusters are. Matilda trains on the cell types it is scored against. Higher is better for every metric.

<details>
<summary>Details</summary>

`params={"Method": {"key": value}}` sets a method's parameters, and `mtb.params_for` lists them. Many methods take none.

`mtb.load_run("out/D22mini")` reloads these results later without running anything.

</details>

```python
res = mtb.run("D22mini", "vertical", methods=METHODS,
                  out_dir="out/D22mini")
res.summary
```

## 4. Plot

`res.plot()` draws the scores as a bubble table. The fills and Overall bars rank the rows within each column, so a small gap can look large: `res.summary` has the values.

<details>
<summary>Details</summary>

Metrics are grouped by family: blue for clustering. Each family starts with an `Overall` bar: each row's mean rank over the family's metrics, scaled so that the lowest gets the shortest bar.

A column with fewer than two different values is drawn grey. A figure of one method is grey everywhere.

</details>

```python
res.plot()
```

## 5. Your own data

Your data needs raw counts for RNA and protein (ADT) in one `.h5ad` file, an ATAC peak matrix of the same cells in another, and a cell type for each cell.

Starting from something else? See the recipes for [a 10x `.h5` file](https://dsichang.github.io/scMultiBench/tutorials/run/#recipe-10x-h5), [labels in a CSV file](https://dsichang.github.io/scMultiBench/tutorials/run/#recipe-labels-csv), [no labels](https://dsichang.github.io/scMultiBench/tutorials/run/#recipe-no-labels) and [several donors or samples](https://dsichang.github.io/scMultiBench/tutorials/run/#recipe-donors).

<details>
<summary>Details: export</summary>

The names in quotes say where your file holds each part. `"X"` is the main matrix, `"obsm:protein"` the matrix `obsm["protein"]`, `"layer:counts"` the layer `layers["counts"]`, and `"obs:celltype"` the column `obs["celltype"]`. A wrong name after `obsm:`, `layer:` or `obs:` raises an error that lists the names the file has.

The example keeps the protein names in `adata.uns["protein_names"]`. A DataFrame in `adata.obsm["protein"]` brings its own names in its columns.

Without `methods=`, `scan` lists every method that reads the modalities in the folder.

</details>

So that the notebook runs as it is, the next cell writes two small example files, `my_file.h5ad` and `my_atac.h5ad`, from 60% of the cells of `D22mini`.

```python
mtb.data.write_example("D22mini")
```

Read your files. For your own data, change every file name marked `# <-`. On Colab, upload the files first: click the folder icon on the left, then the upload icon.

```python
import anndata as ad

adata = ad.read_h5ad("my_file.h5ad")   # <- your RNA and protein file
atac = ad.read_h5ad("my_atac.h5ad")    # <- your ATAC file, the same cells
print(adata)
print(atac)
```

Write the dataset folder. Change each name in quotes marked `# <-`. The printout above lists the names in your files. Write a name listed after `obs:` with that prefix, as `"obs:celltype"`, and one after `obsm:` as `"obsm:protein"`.

```python
NAME = "MYTRI"   # <- a new name for your own data
mtb.io.export_dataset(adata, "mydata/" + NAME,
                      rna="X",                # <- where the RNA counts are:
                                              #    "X" or "layer:counts"
                      adt="obsm:protein",     # <- where the proteins (ADT) are
                      atac=atac,              # the ATAC file, peak counts in X
                      atac_kind="peak",       # the methods here read peaks
                      labels="obs:celltype",  # <- the cell-type column
                      overwrite=True)
mtb.scan("mydata/" + NAME, "vertical", methods=METHODS)
```

```python
mine = mtb.run("mydata/" + NAME, "vertical", methods=METHODS,
                   out_dir="out/" + NAME)
mine.summary
```

```python
mine.plot()
```

## Troubleshooting

`res.summary[res.summary.status != "scored"]` lists each method that failed or was skipped. Its `reason` column ends with the method's error output.

<details>
<summary>Details</summary>

| symptom | fix |
|---|---|
| a method fails on a small dataset | `mtb.scan` names what the method needs from the data's size in its `caveat` column |
| a method times out | raise `timeout=` in `run` |
| batch metrics use the wrong batches | `res.rescore(batch=my_vector)` scores again without running the methods |

</details>
