# Mosaic integration: Mixed, with shared modality, every method

This notebook runs every method that has a variant for mosaic Mixed, with shared modality on `D46mini`: Multigrate, StabMap, UINMF and scMoMaT. The [tutorial](https://dsichang.github.io/scMultiBench/tutorials/mosaic_shared/) explains each step and runs StabMap and UINMF only.

Methods run on Linux. The environments are 3 envs, 2.5 GB on a computer without a GPU and 5.9 GB on one with a GPU.

## 1. Install

The pins keep the numpy and pandas already installed, so Colab needs no restart. On Colab, choose a T4 GPU runtime first (Runtime -> Change runtime type), because changing it later loses the downloads.

```python
import numpy, pandas
pins = f"numpy=={numpy.__version__} pandas=={pandas.__version__}"
%pip install -q "multibench-sc>=0.5.2" {pins}
```

## 2. Download the data and the environments

```python
import multibench as mtb
import pandas as pd

pd.set_option("display.max_columns", None)   # show every column of the tables

mtb.data.fetch("D46mini")
METHODS = ["Multigrate", "StabMap", "UINMF", "scMoMaT"]
envs = mtb.env.install(METHODS)
```

## 3. Run the methods

```python
res = mtb.run("D46mini", "mosaic", methods=METHODS, out_dir="out/D46mini_all")
res.summary
```

## 4. Plot

`res.plot()` draws the scores as a bubble table. Circle size ranks the rows within each column and colour shows the value scaled per column, so a small gap can look large. `res.summary` has the values.

```python
res.plot()
```
