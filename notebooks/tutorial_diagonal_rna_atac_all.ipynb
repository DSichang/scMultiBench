# Diagonal integration: [RNA, ATAC], every method

This notebook runs every method that has a variant for diagonal [RNA, ATAC] on `D27mini`: Conos, GLUE, MultiMAP, Portal, SCALEX, Seurat_v3, Seurat_v5, VIPCCA, iNMF, online_iNMF, scBridge, scJoint, sciCAN and uniPort. The [tutorial](https://dsichang.github.io/scMultiBench/tutorials/diagonal_rna_atac/) explains each step and runs iNMF and online_iNMF only.

Methods run on Linux. The environments are 8 envs, 13.9 GB on a computer without a GPU and 13.2 GB on one with a GPU.

## 1. Install

This cell installs `multibench-sc`. On Colab, choose a T4 GPU runtime first (Runtime -> Change runtime type), because changing it later loses the downloads.

```python
import numpy, pandas
pins = f"numpy=={numpy.__version__} pandas=={pandas.__version__}"
%pip install -q "multibench-sc>=0.5.0" {pins}
```

## 2. Download the data and the environments

```python
import multibench as mtb

mtb.data.fetch("D27mini")
METHODS = ["Conos", "GLUE", "MultiMAP", "Portal", "SCALEX", "Seurat_v3", "Seurat_v5",
           "VIPCCA", "iNMF", "online_iNMF", "scBridge", "scJoint", "sciCAN", "uniPort"]
envs = mtb.env.install(METHODS)
```

## 3. Run the methods

```python
res = mtb.run("D27mini", "diagonal", methods=METHODS,
                  out_dir="out/D27mini_all")
res.summary
```

## 4. Plot

`res.plot()` draws the scores as a bubble table. The fills and Overall bars rank the rows within each column, so a small gap can look large: `res.summary` has the values.

```python
res.plot()
```
