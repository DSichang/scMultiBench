# Vertical integration: RNA + ATAC, every method

This notebook runs every method that has a variant for vertical RNA + ATAC on `D27mini_vertical`: MIRA, MOFA2, Matilda, Multigrate, Seurat_WNN, UINMF, UnitedNet, VIMCCA, iPOLNG, moETM, scMDC, scMM, scMVP and scMoMaT. The [tutorial](https://dsichang.github.io/scMultiBench/tutorials/vertical_rna_atac/) explains each step and runs scMM and VIMCCA only.

Methods run on Linux. The 13 environments download 31.8 GB on a computer without a GPU and 37.6 GB on one with a GPU.

## 1. Install

pip may then print a note about restarting the kernel: no restart is needed, because the pins keep the numpy and pandas already installed. On Colab, choose a T4 GPU runtime first (Runtime -> Change runtime type), because changing it later loses the downloads.

```python
import numpy, pandas
pins = f"numpy=={numpy.__version__} pandas=={pandas.__version__}"
%pip install -q "multibench-sc>=0.6.0" {pins}
```

## 2. Download the data and the environments

```python
import multibench as mtb

mtb.data.fetch("D27mini_vertical")
METHODS = ["MIRA", "MOFA2", "Matilda", "Multigrate", "Seurat_WNN", "UINMF",
           "UnitedNet", "VIMCCA", "iPOLNG", "moETM", "scMDC", "scMM", "scMVP",
           "scMoMaT"]
envs = mtb.env.install(METHODS)
```

## 3. Run the methods

MIRA and Seurat_WNN return a neighbour graph, not an embedding. The package scores embeddings, so their rows have the status `ran` and no scores, and the figure leaves them out.

```python
res = mtb.run("D27mini_vertical", "vertical", methods=METHODS,
              out_dir="out/D27mini_vertical_all")
res.summary
```

## 4. Plot

`res.plot()` draws the scores as a bubble table. Circle size ranks the rows within each column and colour shows the value scaled per column, so a small gap can look large. `res.summary` has the values.

```python
res.plot()
```
