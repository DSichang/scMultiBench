# Cross integration: Multiple ADT + ATAC, every method

This notebook runs every method that has a variant for cross Multiple ADT + ATAC on `D58mini`: MOFA2, Multigrate, StabMap, UINMF and scMoMaT. The [tutorial](https://dsichang.github.io/scMultiBench/tutorials/cross_adt_atac/) explains each step and runs StabMap and UINMF only.

Methods run on Linux. The environments are 4 envs, 15.8 GB on a computer without a GPU and 15.5 GB on one with a GPU.

## 1. Install

This cell installs `multibench-sc`. It needs no restart. On Colab, choose a T4 GPU runtime first (Runtime -> Change runtime type), because changing it later loses the downloads.

```python
import numpy, pandas
pins = f"numpy=={numpy.__version__} pandas=={pandas.__version__}"
%pip install -q "multibench-sc>=0.5.0" {pins}
```

## 2. Download the data and the environments

```python
import multibench as mtb

mtb.data.fetch("D58mini")
METHODS = ["MOFA2", "Multigrate", "StabMap", "UINMF", "scMoMaT"]
envs = mtb.env.install(METHODS)
```

## 3. Run the methods

```python
res = mtb.run("D58mini", "cross", methods=METHODS,
                  out_dir="out/D58mini_all")
res.summary
```

## 4. Plot

`res.plot()` draws the scores as a bubble table. Each column is scaled on its own: its lightest fill is that column's lowest value, not zero.

```python
res.plot()
```
