# Diagonal integration: [multiple RNA, multiple ATAC], every method

This notebook runs every method that has a variant for diagonal [multiple RNA, multiple ATAC] on `D37mini`: GLUE, iNMF, online_iNMF and scJoint. The [tutorial](https://dsichang.github.io/scMultiBench/tutorials/diagonal_multi/) explains each step and runs iNMF and online_iNMF only.

Methods run on Linux. The environments are 3 envs, 4.8 GB on a computer without a GPU and 8.3 GB on one with a GPU.

## 1. Install

This cell installs `multibench-sc`. On Colab, choose a T4 GPU runtime first (Runtime -> Change runtime type), because changing it later loses the downloads.

```python
import numpy, pandas
pins = f"numpy=={numpy.__version__} pandas=={pandas.__version__}"
%pip install -q "multibench-sc>=0.5.2" {pins}
```

## 2. Download the data and the environments

```python
import multibench as mtb
import pandas as pd

pd.set_option("display.max_columns", None)   # show every column of the tables

mtb.data.fetch("D37mini")
METHODS = ["GLUE", "iNMF", "online_iNMF", "scJoint"]
envs = mtb.env.install(METHODS)
```

## 3. Run the methods

```python
res = mtb.run("D37mini", "diagonal", methods=METHODS,
              out_dir="out/D37mini_all")
res.summary
```

## 4. Plot

`res.plot()` draws the scores as a bubble table. Circle size ranks the rows within each column and colour shows the value scaled per column, so a small gap can look large: `res.summary` has the values.

```python
res.plot()
```
